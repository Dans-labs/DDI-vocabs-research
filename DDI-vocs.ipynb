## DDI Vocabularies RDF

As mentioned above **CESSDA Vocabularies API returns metadata on the vocabularies and their terms, NOT RDF.** Although the concepts are described to certain extend, as you can see in [voc_AggregationMethod.json](voc_AggregationMethod.json) and the extract below, it is a JSON metadata and **it is not the RDF + SKOS which Skomos and other semantic system require**.

```json
{
    "id": 18758,
    "uri": "http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/7975ed0",
    "notation": "ArithmeticMean",
    "title": "Arithmetic mean",
    "definition": "Mathematical average of a set of values. The mean is calculated by adding up two or more values and dividing the total by their number. In social/political science, it is usually the sum of the measurements divided by the number of subjects, or cases.",
    "previousConcept": 8203,
    "introducedInVersionId": 268,
    "position": 1,
    "deprecated": false,
    "validFrom": "2019-11-25"
}
```

In Order to obtain the RDF we can use CESSDA or DDI API. 

### CESSDA API to Download RDF

<span style="color:orange">Note that although we see this method exposed in CESSDA Vocabulary Service website (allowing use to download the RDF), this method is not clearly documented in [User Guide](https://api.tech.cessda.eu/).

<span style="color:green">Unlike the DDI API, **CESSDA API responds with a correctly formed, with Concepts belonging to the skos:Concept class**</span>

**See the vocabularies in [DDI-vocs/](DDI-vocs/) directory**




### ☠️ DDI API to Download RDF ☠️

In order to obtain the RDF we have to follow the instructions in https://documenter.getpostman.com/view/2085804/2s8ZDVYiHx

for instances: Get latest version of a CV by name https://rdf-vocabulary.ddialliance.org/cv/:name with supported values: application/ld+json, application/n-triples, application/rdf+xml, text/turtle


Below I requested the RDF (ttl) content of the AggregationMethod vocabulary and save it to [DDI-AggregationMethod.ttl](DDI-AggregationMethod.ttl)

The file includes a `skos:ConceptScheme`, which refers to the vocabularies top concepts. Which is how it is supposed to be.

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/>
        rdf:type             skos:ConceptScheme ;
        ...
        skos:hasTopConcept   <http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61> , <http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>, ... ;
        ...
        skos:prefLabel       "Aggregeringsmetode"@no , "Aggregation Method"@en .
```

However,there is a  <span style="color:red">**Issue 2: with those top Concepts, they do not member of the skos:Concept class or any other class, that would be able to make them a concept. They are simplya URI with a label. This breaks the fundamentals of RDF and SKOS and needs to be addressed**</span>

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61>
        skos:prefLabel  "Modus"@no , "Mode"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/bdc1c84>
        skos:prefLabel  "Persentil"@no , "Percentile rank"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>
        skos:prefLabel  "Annet"@no , "Other"@en .
```

<span style="color:green">**The Correct way would be:**</span> to include `rdf:type skos:Concept` to each concept:

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61>
        rdf:type             skos:Concept;
        skos:prefLabel  "Modus"@no , "Mode"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/bdc1c84>
        rdf:type             skos:Concept;
        skos:prefLabel  "Persentil"@no , "Percentile rank"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>
        rdf:type             skos:Concept;
        skos:prefLabel  "Annet"@no , "Other"@en .

```

````python
import requests
import json 
import rdflib
from pathlib import Path
from pprint import pprint
````

````python
vocs = requests.get("https://vocabularies.cessda.eu/v2/vocabularies-published") # overview of all vocs vocabularies.cessda.eu
vocs_dict = json.loads(vocs.text)
````

````python
# DDI Alliance vocs
vocs_dict = json.loads(vocs.text)
ddi_vocs = vocs_dict["DDI Alliance"]
ddi_vocs_name = ddi_vocs.keys()
print(f'total of: {len(ddi_vocs_name)} DDI Alliance vocs')
print(f'vocs: {list(ddi_vocs_name)}')

ddi_vocs_api = [] # API endpoints for the latest version of each DDI voc

for voc_name, voc_info in ddi_vocs.items():
    en_vocs = ddi_vocs[voc_name]['en(SL)']
    last_version = sorted(en_vocs.keys())[-1]
    last_version_eng_voc = en_vocs[last_version]
    print(last_version_eng_voc)
    ddi_vocs_api.append(f'https://vocabularies.cessda.eu/{last_version_eng_voc}')
    # print(f'voc: {voc_name}, langs: {", ".join(ddi_vocs[voc_name].keys())}')
````

````python
for i, voc_api in enumerate(ddi_vocs_api):
    if i < 1: 
        voc = requests.get(voc_api)
        voc_dict = json.loads(voc.text)
        print(f'voc: {voc_dict["notation"]}, id: {voc_dict["uri"]}')
        open(f'voc_{voc_dict["notation"]}.json', 'w').write(json.dumps(voc_dict, indent=4))

````

````python
# GET 
# https://vocabularies.cessda.eu/v2/vocabularies/AggregationMethod/1.1.2?languageVersion=en-1.1.2

# curl --location 'https://vocabularies.cessda.eu/v2/vocabularies/AggregationMethod/1.1.2?languageVersion=en-1.1.2' --header 'Accept: '

request_headers = {'Accept': 'application/rdf+xml'}
response = requests.get('https://vocabularies.cessda.eu/v2/vocabularies/AggregationMethod/1.1.2', headers=request_headers)
# save response to CESSDA-DDI-AggregationMethod.rdf
with open('DDI-vocs/CESSDA-DDI-AggregationMethod.rdf', 'w') as f:
    f.write(response.text)
# make response.text onto a graph 
cessda_ddi_aggm_graph = rdflib.Graph()
cessda_ddi_aggm_graph.parse(data=response.text, format='application/rdf+xml')
# export graph to turtle file: CESSDA-DDI-AggregationMethod.ttl
cessda_ddi_aggm_graph.serialize(destination='DDI-vocs/CESSDA-DDI-AggregationMethod.ttl', format='turtle')
````

````python
# Download all DDI vocs as RDF and TTL to 'DDI-vocs/' dir

request_headers = {'Accept': 'application/rdf+xml'}

for voc_name, voc_info in ddi_vocs.items():
    en_vocs = ddi_vocs[voc_name]['en(SL)']
    last_version = sorted(en_vocs.keys())[-1]
    # print(voc_name, last_version)
    download_url = f'https://vocabularies.cessda.eu/v2/vocabularies/{voc_name}/{last_version}'
    print(download_url)
    response = requests.get(download_url, headers=request_headers)
    # save RDF(XML)
    with open(f'DDI-vocs/DDI-{voc_name}.rdf', 'w') as f:
        f.write(response.text)
    # save convesiont to turtle
    graph = rdflib.Graph()
    graph.parse(data=response.text, format='application/rdf+xml')
    # export graph to turtle file: DDI-{voc_name}.ttl
    graph.serialize(destination=f'DDI-vocs/DDI-{voc_name}.ttl', format='turtle')
````

````python
# use SPARQL query CESSDA-DDI-AggregationMethod.rdf or CESSDA-DDI-AggregationMethod.ttl for skos:ConceptScheme and skos:Concepts instances
# both lists should be the same
q_concepts_from_conceptSchema = """
    PREFIX skos: <http://www.w3.org/2004/02/skos/core#>
    SELECT ?concept
    WHERE {
    ?conceptScheme a skos:ConceptScheme ;
        skos:hasTopConcept ?concept .
    }
"""

q_concepts = """
    PREFIX skos: <http://www.w3.org/2004/02/skos/core#>
    SELECT ?concept
    WHERE {
    ?concept a skos:Concept .
    }
"""


for voc in Path('DDI-vocs/').glob('DDI-*.rdf'):
    graph = rdflib.Graph()
    graph.parse(voc, format='application/rdf+xml')
    # print(f'-- voc: {voc} --')
    resp_concepts_from_conceptSchema = graph.query(q_concepts_from_conceptSchema)
    resp_concepts = graph.query(q_concepts)
    uris_topConcepts_of_conceptSchema = sorted([row[0].n3() for row in resp_concepts_from_conceptSchema ])
    uris_concepts = sorted([row[0].n3() for row in resp_concepts])
    if uris_topConcepts_of_conceptSchema != uris_concepts:
        print(f'WARNING {voc} has different lists of skos:hasTopConcept values and skos:Concept')    
    # assert uris_topConcepts_of_conceptSchema == uris_concepts, "The lists of top concepts and concepts are not the same."
````

````python
# investigate discrepancies between skos:hasTopConcept values and skos:Concept instances in DDI vocs 
# case study: DDI-vocs/DDI-ContributorRole.rdf

graph = rdflib.Graph()
graph.parse('DDI-vocs/DDI-ContributorRole.rdf', format='application/rdf+xml')
# print(graph.serialize(format='turtle'))

# query: do all skos:hasTopConcept values have a corresponding skos:Concept instance in the graph?
q_concepts_from_conceptSchema = """
    PREFIX skos: <http://www.w3.org/2004/02/skos/core#>
    SELECT ?concept
    WHERE {
    ?conceptScheme a skos:ConceptScheme ;
        skos:hasTopConcept ?concept .
    }
"""
# query: what are the instances of skos:Concept which are not values of skos:hasTopConcept from skos:TopConcept?
q_concepts_not_topConcept = """
PREFIX skos: <http://www.w3.org/2004/02/skos/core#>

SELECT DISTINCT ?concept
WHERE {
  ?concept a skos:Concept .

  FILTER NOT EXISTS {
    ?scheme a skos:ConceptScheme ;
            skos:hasTopConcept ?concept .
  }
}
ORDER BY ?concept
"""

r_concepts_not_topConcept = graph.query(q_concepts_not_topConcept)
for row in r_concepts_not_topConcept:
    print(row)

# TODO testing
````

````python
# join all ConceptSchemes and Concepts from all DDI vocs into one graph 
# and export to RDF/XMLs
combined_graph = rdflib.Graph()
for voc in Path('DDI-vocs/').glob('DDI-*.rdf'):
    graph = rdflib.Graph()
    graph.parse(voc, format='application/rdf+xml')
    combined_graph += graph
combined_graph.serialize('combined_graph.rdf', format='application/rdf+xml')
combined_graph.serialize('combined_graph.ttl', format='turtle')

# TODO: 
# 1) query skos:ConceptScheme and skos:Concept instances with subset language labels
# 2) include metadata about the voc 
# 3) view in Skosmos
# see AATCss
````
