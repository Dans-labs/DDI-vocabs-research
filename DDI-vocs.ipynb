## DDI Vocabularies RDF

As mentioned above **CESSDA Vocabularies API only returns metadata on the vocabularies and their terms, NOT RDF.** Although the concepts are described to certain extend, as you can see in [voc_AggregationMethod.json](voc_AggregationMethod.json) and the extract below, it is a JSON metadata and **it is not the RDF + SKOS which Skomos and other semantic system require**.

```json
{
    "id": 18758,
    "uri": "http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/7975ed0",
    "notation": "ArithmeticMean",
    "title": "Arithmetic mean",
    "definition": "Mathematical average of a set of values. The mean is calculated by adding up two or more values and dividing the total by their number. In social/political science, it is usually the sum of the measurements divided by the number of subjects, or cases.",
    "previousConcept": 8203,
    "introducedInVersionId": 268,
    "position": 1,
    "deprecated": false,
    "validFrom": "2019-11-25"
}
```

In order to obtain the RDF we have to follow the instructions in https://documenter.getpostman.com/view/2085804/2s8ZDVYiHx

for instances: Get latest version of a CV by name https://rdf-vocabulary.ddialliance.org/cv/:name with supported values: application/ld+json, application/n-triples, application/rdf+xml, text/turtle


### AggregationMethod RDF (ttl)

Below I requested the RDF (ttl) content of the AggregationMethod vocabulary and save it to (DDI-AggregationMethod.ttl)[DDI-AggregationMethod.ttl]

The file includes a `skos:ConceptScheme`, which refers to the vocabularies top concepts. Which is how it is supposed to be.

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/>
        rdf:type             skos:ConceptScheme ;
        ...
        skos:hasTopConcept   <http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61> , <http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>, ... ;
        ...
        skos:prefLabel       "Aggregeringsmetode"@no , "Aggregation Method"@en .
```

However,there is a ☠️ <span style="color:red">**Issue 2: with those top Concepts, they do not member of the skos:Concept class or any other class, that would be able to make them a concept. They are simplya URI with a label. This breaks the fundamentals of RDF and SKOS and needs to be addressed**</span> ☠️

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61>
        skos:prefLabel  "Modus"@no , "Mode"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/bdc1c84>
        skos:prefLabel  "Persentil"@no , "Percentile rank"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>
        skos:prefLabel  "Annet"@no , "Other"@en .
```

<span style="color:green">**The Correct way would be:**</span> to include `rdf:type skos:Concept` to each concept:

```
<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/650be61>
        rdf:type             skos:Concept;
        skos:prefLabel  "Modus"@no , "Mode"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/bdc1c84>
        rdf:type             skos:Concept;
        skos:prefLabel  "Persentil"@no , "Percentile rank"@en .

<http://rdf-vocabulary.ddialliance.org/cv/AggregationMethod/1.1.2/6311ae1>
        rdf:type             skos:Concept;
        skos:prefLabel  "Annet"@no , "Other"@en .

```

````python
import requests
import json 
from pprint import pprint
````

````python
vocs = requests.get("https://vocabularies.cessda.eu/v2/vocabularies-published") # overview of all vocs vocabularies.cessda.eu
vocs_dict = json.loads(vocs.text)
````

````python
# DDI Alliance vocs
vocs_dict = json.loads(vocs.text)
ddi_vocs = vocs_dict["DDI Alliance"]
ddi_vocs_name = ddi_vocs.keys()
print(f'total of: {len(ddi_vocs_name)} DDI Alliance vocs')
print(f'vocs: {list(ddi_vocs_name)}')

ddi_vocs_api = [] # API endpoints for the latest version of each DDI voc

for voc_name, voc_info in ddi_vocs.items():
    en_vocs = ddi_vocs[voc_name]['en(SL)']
    last_version = sorted(en_vocs.keys())[-1]
    last_version_eng_voc = en_vocs[last_version]
    print(last_version_eng_voc)
    ddi_vocs_api.append(f'https://vocabularies.cessda.eu/{last_version_eng_voc}')
    # print(f'voc: {voc_name}, langs: {", ".join(ddi_vocs[voc_name].keys())}')
````

````python
for i, voc_api in enumerate(ddi_vocs_api):
    if i < 1: 
        voc = requests.get(voc_api)
        voc_dict = json.loads(voc.text)
        print(f'voc: {voc_dict["notation"]}, id: {voc_dict["uri"]}')
        open(f'voc_{voc_dict["notation"]}.json', 'w').write(json.dumps(voc_dict, indent=4))

````

````python
# GET the AggregationMethod DDI vocabulary and save to AggregationMethod.ttl
# same as curl --location 'https://rdf-vocabulary.ddialliance.org/cv/AggregationMethod' --header 'Accept: text/turtle' 

request_headers = {'Accept': 'text/turtle'}
response = requests.get('https://rdf-vocabulary.ddialliance.org/cv/AggregationMethod', headers=request_headers)
print(response.text)
with open('DDI-AggregationMethod.ttl', 'w') as f:
    f.write(response.text)
````
